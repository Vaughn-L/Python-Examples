```python
#@title Cell 0 - Set Known Values
#Set Known Values
alpha = .0056
S0 = .45
p1 = 1360
p2 = 1187
p3 = 1278
p4 = 1489

q1_avg = 4005
q2_avg = 4637
q3_avg = 4478
q4_avg = 1124

#Pick Which Firms Will Merge:
merging_firmA = "1"
merging_firmB = "2"
```

```python
#@title Cell 1 - Problem Set 2
class color:
   PURPLE = '\033[95m'
   CYAN = '\033[96m'
   DARKCYAN = '\033[36m'
   BLUE = '\033[94m'
   GREEN = '\033[92m'
   YELLOW = '\033[93m'
   RED = '\033[91m'
   BOLD = '\033[1m'
   UNDERLINE = '\033[4m'
   END = '\033[0m'

#Calculate Market Size
Market = (q1_avg+q2_avg+q3_avg+q4_avg)/(1-S0)
q0_avg = Market -(q1_avg+q2_avg+q3_avg+q4_avg)

#Calculate Market Shares
share1 = q1_avg / Market
share2 = q2_avg / Market
share3 = q3_avg / Market
share4 = q4_avg / Market

#Set Heading
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BOLD +'PART 1:RECOVER KEY PARAMETERS'+ color.END))
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'MARKET SIZE'+ color.END))

#Generate Output for User
print("The Market size is about "+str(round(Market,2))+" units")
print("The outside quantity is about "+str(round(q0_avg,2))+" units")

#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'MARKET SHARES'+ color.END))

#Generate Output for User
print("Firm 1 market share: "+str((round(share1,3)*100))+"%")
print("Firm 2 market share: "+str((round(share2,3)*100))+"%")
print("Firm 3 market share: "+str((round(share3,2)*100))+"%")
print("Firm 4 market share: "+str((round(share4,3)*100))+"%")

#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'MARGINAL COSTS'+ color.END))

#Estimate Costs
c1 = -1 * ((p1/(alpha * p1 * (1-share1)))-p1 )
c2 = -1 * ((p2/(alpha * p2 * (1-share2)))-p2 )
c3 = -1 * ((p3/(alpha * p3 * (1-share3)))-p3 )
c4 = -1 * ((p4/(alpha * p4 * (1-share4)))-p4 )

#Generate Output for User
print("Estimated Firm 1 Cost: "+str(round(c1,2)))
print("Estimated Firm 2 Cost: "+str(round(c2,2)))
print("Estimated Firm 3 Cost: "+str(round(c3,2)))
print("Estimated Firm 4 Cost: "+str(round(c4,2)))

# Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'PRODUCT VALUES'+ color.END))

#Estimate Values
from scipy.optimize import fsolve
import numpy as np

# Define the equations
def equations(variables,share1,share2,share3,share4):
    v1, v2, v3, v4 = variables
    Denom = (1 + np.exp(v1 - (alpha*p1)) + np.exp(v2 - (alpha*p2)) + np.exp(v3 - (alpha*p3)) + np.exp(v4 - (alpha*p4)))
    eq1 = share1 - np.exp(v1 - (alpha*p1)) / Denom
    eq2 = share2 - np.exp(v2 - (alpha*p2)) / Denom
    eq3 = share3 - np.exp(v3 - (alpha*p3)) / Denom
    eq4 = share4 - np.exp(v4 - (alpha*p4)) / Denom
    return eq1, eq2, eq3, eq4

# Solve the equations - Maximization Approach
variables = fsolve(equations, (3, 3, 3, 3), args=(share1,share2,share3,share4))

#Equations to Estimate Values without maximization approach:
v1 = np.log(share1/S0) + (alpha*p1)
v2 = np.log(share2/S0) + (alpha*p2)
v3 = np.log(share3/S0) + (alpha*p3)
v4 = np.log(share4/S0) + (alpha*p4)

# Generate Output for User
print("Estimate of Product 1 Value: ", variables[0])
print("Estimate of Product 2 Value: ", variables[1])
print("Estimate of Product 3 Value: ", variables[2])
print("Estimate of Product 4 Value: ", variables[3])

#Set Heading
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BOLD +'PART 2: Four Firm Monopolist'+ color.END))
print('--------------------------------------------------------------------------------------------------------------------')
#print('{:^120s}'.format(color.BLUE +'DIVERSION RATIO ESTIMATES'+ color.END))

#Generate Variables containing the value of the Diversion Ratio for each combination of firms
Firms = ["1","2","3","4"]
for firm in Firms:
  for X in Firms:
    d_ratio = "d"+X+firm
    y = "share"+str(firm)
    z = "share"+str(X)
    exec(d_ratio + "=" + y + "/(1-"+z+")")

  # Generate Output for User
    #print("Estimated Diversion Ratio "+d_ratio+": "+str(locals()[d_ratio]))

# Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'Generate K values'+ color.END))

#Generate Variables containing the value of K for each firm
k_1 = np.exp(v1 - alpha * c1)
k_2 = np.exp(v2 - alpha * c2)
k_3 = np.exp(v3 - alpha * c3)
k_4 = np.exp(v4 - alpha * c4)
K = k_1 + k_2 + k_3 + k_4

#Generate Output for User
print('Firm 1 k Estimate: '+str(round(k_1,2)))
print('Firm 2 k Estimate: '+str(round(k_2,2)))
print('Firm 3 k Estimate: '+str(round(k_3,2)))
print('Firm 4 k Estimate: '+str(round(k_4,2)))
print('K Estimate: '+str(round(K,2)))

# Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'Solve for M and Price under Monopoly'+ color.END))

#Solve for m
import math
def f(m, K):
    return (m-1) * math.exp(m) - K

def f_prime(m):
    return m * math.exp(m)

def solve(m_guess, K, tol=1e-6, max_iters=100):
    m = m_guess
    for i in range(max_iters):
        f_val = f(m, K)
        f_prime_val = f_prime(m)
        m_next = m - f_val / f_prime_val
        if abs(m_next - m) < tol:
            return m_next
        m = m_next
    raise ValueError("Failed to converge")

# example usage:
m = solve(1.0, K)

#Generate price under monopoly for each firm using m
P_Monopoly1 = c1 + m /alpha
P_Monopoly2 = c2 + m /alpha
P_Monopoly3 = c3 + m /alpha
P_Monopoly4 = c4 + m /alpha

#Generate Output for User
print("Monopolist Price, Firm 1: "+str(round(P_Monopoly1,2)))
print("Monopolist Price, Firm 2: "+str(round(P_Monopoly2,2)))
print("Monopolist Price, Firm 3: "+str(round(P_Monopoly3,2)))
print("Monopolist Price, Firm 4: "+str(round(P_Monopoly4,2)))

#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'Price Changes'+ color.END))

#Generate variables containing value of absolute changes in price for each firm + weighted average
Price_change1_abs = P_Monopoly1 - p1
Price_change2_abs = P_Monopoly2 - p2
Price_change3_abs = P_Monopoly3 - p3
Price_change4_abs = P_Monopoly4 - p4
Limited_Market = Market - q0_avg
Weighted_Price_Pre = (q1_avg/Limited_Market)*p1 + (q2_avg/Limited_Market)*p2 + (q3_avg/Limited_Market)*p3+ (q4_avg/Limited_Market)*p4
Weighted_Price_Post = (q1_avg/Limited_Market)*P_Monopoly1 + (q2_avg/Limited_Market)*P_Monopoly2 + (q3_avg/Limited_Market)*P_Monopoly3+ (q4_avg/Limited_Market)*P_Monopoly4
Price_change_weighted_abs = Weighted_Price_Post - Weighted_Price_Pre

#Generate variables containing value of percentage changes in price for each firm + weighted average
Price_change1_pct = Price_change1_abs / p1
Price_change2_pct = Price_change2_abs / p2
Price_change3_pct = Price_change3_abs / p3
Price_change4_pct = Price_change4_abs / p4
Price_change_weighted_pct = Price_change_weighted_abs / Weighted_Price_Pre

#Generate Output for User
print("Firm 1 Price Increase: +$"+str(round(Price_change1_abs,2))+" | ("+str(round(Price_change1_pct*100,2))+"%)")
print("Firm 2 Price Increase: +$"+str(round(Price_change2_abs,2))+" | ("+str(round(Price_change2_pct*100,2))+"%)")
print("Firm 3 Price Increase: +$"+str(round(Price_change3_abs,2))+" | ("+str(round(Price_change3_pct*100,2))+"%)")
print("Firm 4 Price Increase: +$"+str(round(Price_change4_abs,2))+" | ("+str(round(Price_change4_pct*100,2))+"%)")
print("")
print("Weighted Average Price Increase: +$"+str(round(Price_change_weighted_abs,2))+" | ("+str(round(Price_change_weighted_pct*100,2))+"%)")

#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BOLD + 'PART 3: Merger Between Firm '+merging_firmA+' and Firm '+merging_firmB + " - Exact Approach"+ color.END))
print('--------------------------------------------------------------------------------------------------------------------')
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'Generate Limited K value'+ color.END))

#Solve for K in two-firm merger
exec("K_limited = k_" + merging_firmA + " + k_"+merging_firmB)

#Generate Output for User
print('K Estimate: '+str(round(K_limited,2)))

# Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'Solve for M and Price under Merger'+ color.END))

#Solve for m_limited
m_limited = solve(1.0, K_limited)

#Generate price post-merger for each firm using m
P_Merger1 = c1 + m_limited /alpha
P_Merger2 = c2 + m_limited /alpha
P_Merger3 = c3 + m_limited /alpha
P_Merger4 = c4 + m_limited /alpha


#Generate Output for User
print("Post-Merger Price, Firm 1: "+str(round(P_Merger1,2)))
print("Post-Merger Price, Firm 2: "+str(round(P_Merger2,2)))
print("Post-Merger Price, Firm 3: "+str(round(P_Merger3,2)))
print("Post-Merger Price, Firm 4: "+str(round(P_Merger4,2)))

#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BLUE +'Merger Between Firm '+merging_firmA+' and Firm'+merging_firmB+' - PRICE CHANGES'+ color.END))

#Generate variables containing value of absolute changes in price for each firm + weighted average
Price_change1_abs_merge = P_Merger1 - p1
Price_change2_abs_merge = P_Merger2 - p2
Price_change3_abs_merge = P_Merger3 - p3
Price_change4_abs_merge = P_Merger4 - p4
Limited_Market_merge = Market - q0_avg
Weighted_Price_Pre_merge = (q1_avg/Limited_Market_merge)*p1 + (q2_avg/Limited_Market_merge)*p2+ (q3_avg/Limited_Market_merge)*p3+(q4_avg/Limited_Market_merge)*p4
Weighted_Price_Post_merge = (q1_avg/Limited_Market_merge)*P_Merger1 + (q2_avg/Limited_Market_merge)*P_Merger2 + (q3_avg/Limited_Market_merge)*P_Merger3 + (q4_avg/Limited_Market_merge)*P_Merger4
Price_change_weighted_abs_merge = Weighted_Price_Post_merge - Weighted_Price_Pre_merge

#Generate variables containing value of percentage changes in price for each firm + weighted average
Price_change1_pct_merge = Price_change1_abs_merge / p1
Price_change2_pct_merge = Price_change2_abs_merge / p2
Price_change3_pct_merge = Price_change3_abs_merge / p3
Price_change4_pct_merge = Price_change4_abs_merge / p4
Price_change_weighted_pct_merge = Price_change_weighted_abs_merge / Weighted_Price_Pre_merge

#Generate Output for User
print("Firm 1 Price Increase: +$"+str(round(Price_change1_abs_merge,2))+" | ("+str(round(Price_change1_pct_merge*100,2))+"%)")
print("Firm 2 Price Increase: +$"+str(round(Price_change2_abs_merge,2))+" | ("+str(round(Price_change2_pct_merge*100,2))+"%)")
print("Firm 3 Price Increase: +$"+str(round(Price_change3_abs_merge,2))+" | ("+str(round(Price_change3_pct_merge*100,2))+"%)")
print("Firm 4 Price Increase: +$"+str(round(Price_change4_abs_merge,2))+" | ("+str(round(Price_change4_pct_merge*100,2))+"%)")
print("")
print("Weighted Average Price Increase: +$"+str(round(Price_change_weighted_abs_merge,2))+" | ("+str(round(Price_change_weighted_pct_merge*100,2))+"%)")

#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BOLD + 'PART 4: Merger Between Firm '+merging_firmA+' and Firm '+merging_firmB + " - MPT-UPP Approach"+ color.END))
print('--------------------------------------------------------------------------------------------------------------------')
print('--------------------------------------------------------------------------------------------------------------------')

#Identify Non-Merging Firms
Firm_List = ["1","2","3","4"]
Firm_List.remove(merging_firmA)
Firm_List.remove(merging_firmB)

#f equations partial derivatives
for firm in Firms:
  for X in Firms:
    f_part = "f_"+firm+"_"+X
    y = "share"+str(firm)
    z = "share"+str(X)
    if firm == X:
      exec(f_part + "= -" + y + "/(1-"+y+") - 1")
    else:
      exec(f_part + "= " + y + "*"+z+"/(1-"+y+")**2")

#Put f partial derivatives into matrix
F_matrix = np.array([[f_1_1,f_1_2,f_1_3,f_1_4],
                     [f_2_1,f_2_2,f_2_3,f_2_4],
                     [f_3_1,f_3_2,f_3_3,f_3_4],
                     [f_4_1,f_4_2,f_4_3,f_4_4]])

#g equations (UPPs)
exec("UPP"+merging_firmA+" = d"+merging_firmA+merging_firmB+"* (p"+merging_firmB+" - c"+merging_firmB+")") # Merging Firm A
exec("UPP"+merging_firmB+" = d"+merging_firmA+merging_firmA+"* (p"+merging_firmA+" - c"+merging_firmA+")") # Merging Firm B
exec("UPP"+Firm_List[0]+" = 0") # First non-merging firm
exec("UPP"+Firm_List[1]+" = 0") # Second non-merging firm

#Put UPPs into matrix
UPP_matrix = np.array([UPP1,UPP2,UPP3,UPP4])

#g equations partial derivatives - Merging Firm A
exec("g_"+merging_firmA+"_"+merging_firmA+" = 0") # 0 with respect to own price
exec("g_"+merging_firmA+"_"+merging_firmB+" = share"+merging_firmB+"*(share"+merging_firmA+" - 1 + alpha*(p"+merging_firmB+" - c"+merging_firmB+")*share"+merging_firmA+"*(share"+merging_firmA+" + share"+merging_firmB+" - 1)) / (share"+merging_firmA+" - 1)**2") # with respect to other merging firm price
exec("g_"+merging_firmA+"_"+Firm_List[0]+" = (p"+merging_firmB+"- c"+merging_firmB+") * (alpha * share"+merging_firmB+"* share"+Firm_List[0]+")/ ((share"+merging_firmA+"-1)**2)") #with respect to first non-merging firm
exec("g_"+merging_firmA+"_"+Firm_List[1]+" = (p"+merging_firmB+"- c"+merging_firmB+") * (alpha * share"+merging_firmB+"* share"+Firm_List[1]+")/ ((share"+merging_firmA+"-1)**2)") # with respect to second non-merging firm

#g equations partial derivatives - Merging Firm B
exec("g_"+merging_firmB+"_"+merging_firmB+" = 0") # 0 with respect to own price
exec("g_"+merging_firmB+"_"+merging_firmA+" = share"+merging_firmA+"*(share"+merging_firmB+" - 1 + alpha*(p"+merging_firmA+" - c"+merging_firmA+")*share"+merging_firmB+"*(share"+merging_firmA+" + share"+merging_firmA+" - 1)) / (share"+merging_firmB+" - 1)**2") # with respect to other merging firm price
exec("g_"+merging_firmB+"_"+Firm_List[0]+" = (p"+merging_firmA+"- c"+merging_firmA+") * (alpha * share"+merging_firmA+"* share"+Firm_List[0]+")/ ((share"+merging_firmB+"-1)**2)") #with respect to first non-merging firm
exec("g_"+merging_firmB+"_"+Firm_List[1]+" = (p"+merging_firmA+"- c"+merging_firmA+") * (alpha * share"+merging_firmA+"* share"+Firm_List[1]+")/ ((share"+merging_firmB+"-1)**2)") # with respect to second non-merging firm

#g equations partial derivatives - First non-merging Firm
exec("g_"+Firm_List[0]+"_"+merging_firmA+" = 0") # with respect to Merging Firm A
exec("g_"+Firm_List[0]+"_"+merging_firmB+" = 0") # with respect to Merging Firm B
exec("g_"+Firm_List[0]+"_"+Firm_List[0]+" = 0") # with respect to first non-merging Firm
exec("g_"+Firm_List[0]+"_"+Firm_List[1]+" = 0") # with respect to second non-merging Firm

#g equations partial derivatives - Second non-merging Firm
exec("g_"+Firm_List[1]+"_"+merging_firmA+" = 0") # with respect to Merging Firm A
exec("g_"+Firm_List[1]+"_"+merging_firmB+" = 0") # with respect to Merging Firm B
exec("g_"+Firm_List[1]+"_"+Firm_List[0]+" = 0") # with respect to first non-merging Firm
exec("g_"+Firm_List[1]+"_"+Firm_List[1]+" = 0") # with respect to second non-merging Firm

#Put g partial derivatives into matrix
G_matrix = np.array([[g_1_1,g_1_2,g_1_3,g_1_4],
                     [g_2_1,g_2_2,g_2_3,g_2_4],
                     [g_3_1,g_3_2,g_3_3,g_3_4],
                     [g_4_1,g_4_2,g_4_3,g_4_4]])

#Make h matrix
H_matrix = F_matrix + G_matrix

#invert, multiply by negative 1 to get cost pass through matrix
from numpy.linalg import inv
H_matrix_inv = (inv(np.matrix(H_matrix)))*-1

#multiply
Part4_Estimate = H_matrix_inv @ UPP_matrix

#find weighted value of absolute change
Part4_Pre_Weight  = (q1_avg/Limited_Market_merge)*p1 + (q2_avg/Limited_Market_merge)*p2+ (q3_avg/Limited_Market_merge)*p3+(q4_avg/Limited_Market_merge)*p4
Part4_Post_Weight = (q1_avg/Limited_Market_merge)*(p1 + Part4_Estimate[0,0]) + (q2_avg/Limited_Market_merge)*(p2 + Part4_Estimate[0,1]) + (q3_avg/Limited_Market_merge)*(p3 + Part4_Estimate[0,2]) + (q4_avg/Limited_Market_merge)*(p4 + Part4_Estimate[0,3])
Part4_Change_Weight = Part4_Post_Weight - Part4_Pre_Weight

#Generate variables containing value of percentage changes in price for each firm + weighted average
Part4_pct_1 = Part4_Estimate[0,0] / p1
Part4_pct_2 = Part4_Estimate[0,1] / p2
Part4_pct_3 = Part4_Estimate[0,2] / p3
Part4_pct_4 = Part4_Estimate[0,3] / p4
Part4_Pct_Weight = Part4_Change_Weight / Part4_Pre_Weight

#Generate Output for User
print("Firm 1 Price Increase: +$"+str(round(Part4_Estimate[0,0],2))+" | ("+str(round(Part4_pct_1 *100,2))+"%)")
print("Firm 2 Price Increase: +$"+str(round(Part4_Estimate[0,1],2))+" | ("+str(round(Part4_pct_2 *100,2))+"%)")
print("Firm 3 Price Increase: +$"+str(round(Part4_Estimate[0,2],2))+" | ("+str(round(Part4_pct_3 *100,2))+"%)")
print("Firm 4 Price Increase: +$"+str(round(Part4_Estimate[0,3],2))+" | ("+str(round(Part4_pct_4 *100,2))+"%)")
print("")
print("Weighted Average Price Increase: +$"+str(round(Part4_Change_Weight,2))+" | ("+str(round(Part4_Pct_Weight*100,2))+"%)")



#Section Break
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BOLD + 'PART 5: Merger Between Firm '+merging_firmA+' and Firm '+merging_firmB + " - CPT-UPP Approach"+ color.END))
print('--------------------------------------------------------------------------------------------------------------------')
print('--------------------------------------------------------------------------------------------------------------------')

#invert F_matrix , multiply by negative 1 to get cost pass through matrix
F_matrix_inv = (inv(np.matrix(F_matrix)))*-1

Part5_Estimate =  F_matrix_inv @ UPP_matrix

#find weighted value of absolute change
Part5_Pre_Weight  = (q1_avg/Limited_Market_merge)*p1 + (q2_avg/Limited_Market_merge)*p2+ (q3_avg/Limited_Market_merge)*p3+(q4_avg/Limited_Market_merge)*p4
Part5_Post_Weight = (q1_avg/Limited_Market_merge)*(p1 + Part5_Estimate[0,0]) + (q2_avg/Limited_Market_merge)*(p2 + Part5_Estimate[0,1]) + (q3_avg/Limited_Market_merge)*(p3 + Part5_Estimate[0,2]) + (q4_avg/Limited_Market_merge)*(p4 + Part5_Estimate[0,3])
Part5_Change_Weight = Part5_Post_Weight - Part5_Pre_Weight

#Generate variables containing value of percentage changes in price for each firm + weighted average
Part5_pct_1 = Part5_Estimate[0,0] / p1
Part5_pct_2 = Part5_Estimate[0,1] / p2
Part5_pct_3 = Part5_Estimate[0,2] / p3
Part5_pct_4 = Part5_Estimate[0,3] / p4
Part5_Pct_Weight = Part5_Change_Weight / Part5_Pre_Weight

#Generate Output for User
print("Firm 1 Price Increase: +$"+str(round(Part5_Estimate[0,0],2))+" | ("+str(round(Part5_pct_1 *100,2))+"%)")
print("Firm 2 Price Increase: +$"+str(round(Part5_Estimate[0,1],2))+" | ("+str(round(Part5_pct_2 *100,2))+"%)")
print("Firm 3 Price Increase: +$"+str(round(Part5_Estimate[0,2],2))+" | ("+str(round(Part5_pct_3 *100,2))+"%)")
print("Firm 4 Price Increase: +$"+str(round(Part5_Estimate[0,3],2))+" | ("+str(round(Part5_pct_4 *100,2))+"%)")
print("")
print("Weighted Average Price Increase: +$"+str(round(Part5_Change_Weight,2))+" | ("+str(round(Part5_Pct_Weight*100,2))+"%)")



#Set Heading
print('--------------------------------------------------------------------------------------------------------------------')
print('{:^120s}'.format(color.BOLD +'PART 6: Discussions'+ color.END))
print('--------------------------------------------------------------------------------------------------------------------')

#Import TextWrap Function
import textwrap

Answer60 = "Generally, there seems to be a fair amount of difference between the various methods. The UPP-based methods appear to under-predict price increases."
Answer61 = "6.1 It may depend on the market. I would feel more comfortable making this assumption in a well-established market that has seen price stability and robust competitive behavior before the merger period -- making it more likely that averages are closer to equilibrium conditions."
Answer62 = "6.2 The merger between T-Mobile and Sprint is interesting. If by combining the companies, there is a reduction in marginal costs, it could make the market more competitive. Perhaps the combined company, with lower marginal costs is able to challenge the power of a dominant firm. I need more experience, exposure, and understanding of available data to know if it is possible to retrospectively study the merger."
Answer63 = "6.3 This approach seems to value simplicity and operating in conditions with minimal data. I would want a greater understanding of data availability and other approaches before confidently determining a way to improve this approach."

wrap_60 = textwrap.wrap(Answer60,width=115)
wrap_61 = textwrap.wrap(Answer61,width=115)
wrap_62 = textwrap.wrap(Answer62,width=115)
wrap_63 = textwrap.wrap(Answer63,width=115)

for line in wrap_60:
  print(line)

for line in wrap_61:
  print(line)

print('')
for line in wrap_62:
  print(line)
print('')
for line in wrap_63:
  print(line)

print("")
print("")
```

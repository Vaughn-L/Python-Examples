```python
import os

# Specify the folder path where the files are located
folder_path = r'\\wdcfs1\DCDATA6\Cases\Active\IRS II (027050)\02 Incoming\2024_0922 - Product Types'

# Loop through each file in the folder
for filename in os.listdir(folder_path):
    # Check if the file name contains an underscore
    if '_' in filename:
        # Split the filename at the first underscore and take the part after it
        new_filename = filename.split('_', 1)[1]
        
        # Construct the full paths to the old and new filenames
        old_file = os.path.join(folder_path, filename)
        new_file = os.path.join(folder_path, new_filename)
        
        # Rename the file
        os.rename(old_file, new_file)

print("Files renamed successfully!")
```

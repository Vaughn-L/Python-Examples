```python
#@title Generate Form_5500 CSVs

#For each year: Import F_5500 and F_5500_SF as one dataframe
Year_Range = [2009,2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021]

#Create Dataset of just correct year
for year in Year_Range:
    #Import Small Plan Files into dataframes
    Small_Chunk = pd.read_csv(LATESTFILEPATH+"f_5500_sf_"+str(year)+"_latest.csv", dtype = Small_Dict, chunksize=100000,na_values='+')
    Small = pd.concat(Small_Chunk)
    print("finished opening Small "+ str(year))

    #Because the small plan file uses slightlty different names for their variables, this section of code changes those names to align with the large plan variable names
    #The next lane strips the prefix SF_ from all of the Small plan variable names
    Small.columns = Small.columns.str.replace("SF_","")
    #The next line renames some variable names that should remain the same, but would cause duplicate variable names if left with their current names
    Small = Small.rename(columns = {'PLAN_NAME':'x_PLAN_NAME', 'PLAN_NUM': 'x_PLAN_NUM', 'SPONS_EIN': 'x_SPONS_EIN', 'ADMIN_NAME':'x_ADMIN_NAME', 'ADMIN_PHONE_NUM':'x_ADMIN_PHONE_NUM', 'PREPARER_PHONE_NUM':'x_PREPARER_PHONE_NUM'})

    #Import Large Plan Files into dataframes
    Large_Chunk = pd.read_csv(ALLFILEPATH+"f_5500_"+str(year)+"_all.csv", dtype = Large_Dict, chunksize=100000, na_values ='+')
    Large = pd.concat(Large_Chunk)
    print("finished opening Large "+str(year))


    #The next two lines pulls the names of all variables in the small and large plan dataframes
    Names_to_change = list(Small)
    Replacement_names = list(Large)
    #This loop checks to see if the small plan variable name is contained within a large plan variable name. If true, the small plan variable name is replaced with the name of the large plan variable name.
    for r_name in Replacement_names:
        for c_name in Names_to_change:
            if c_name in r_name:
                Small.rename(columns = {c_name:r_name}, inplace = True)
    #This next line renames the variables we renamed earlier, so that they go back to their original name (and aren't duplicated variable names so we can concat!)
    Small = Small.rename(columns = {'x_PLAN_NAME': 'PLAN_NAME', 'x_PLAN_NUM': 'PLAN_NUM', 'x_SPONS_EIN': 'SPONS_EIN', 'x_ADMIN_NAME':'ADMIN_NAME', 'x_ADMIN_PHONE_NUM':'ADMIN_PHONE_NUM', 'x_PREPARER_PHONE_NUM':'PREPARER_PHONE_NUM'})

    #This section puts the large and small plan files for the same year into one dataframe, and then exports it as a csv so we can use it later on.
    BaseForm = pd.concat([Large, Small])
    #Free up memory
    del Small_Chunk, Small, Large_Chunk, Large

    print("Compiling as csv - "+str(year))
    Correct_Year = BaseForm[BaseForm['FORM_TAX_PRD'].str.startswith(str(year))==True]
    Correct_Year.to_csv(path+str(year)+"Form_5500.csv", index=False, chunksize=100000)
    print("Exported to csv - "+str(year))

    del Correct_Year, BaseForm

################################################################################
#Append Data Found in Other Year's Dataset to appropriate Data File
################################################################################
print('Sorting Data Files into Correct Years')
for year in Year_Range:
    #Import Small Plan Files into dataframes
    Small_Chunk = pd.read_csv(LATESTFILEPATH+"f_5500_sf_"+str(year)+"_latest.csv", dtype = Small_Dict, chunksize=100000,na_values ='+')
    Small = pd.concat(Small_Chunk)
    print("finished opening Small "+ str(year))

    #Because the small plan file uses slightlty different names for their variables, this section of code changes those names to align with the large plan variable names
    #The next lane strips the prefix SF_ from all of the Small plan variable names
    Small.columns = Small.columns.str.replace("SF_","")
    #The next line renames some variable names that should remain the same, but would cause duplicate variable names if left with their current names
    Small = Small.rename(columns = {'PLAN_NAME':'x_PLAN_NAME', 'PLAN_NUM': 'x_PLAN_NUM', 'SPONS_EIN': 'x_SPONS_EIN', 'ADMIN_NAME':'x_ADMIN_NAME', 'ADMIN_PHONE_NUM':'x_ADMIN_PHONE_NUM', 'PREPARER_PHONE_NUM':'x_PREPARER_PHONE_NUM'})

    #Import Large Plan Files into dataframes
    Large_Chunk = pd.read_csv(ALLFILEPATH+"f_5500_"+str(year)+"_all.csv", dtype = Large_Dict, chunksize=100000,na_values = '+')
    Large = pd.concat(Large_Chunk)
    print("finished opening Large "+str(year))

    #The next two lines pulls the names of all variables in the small and large plan dataframes
    Names_to_change = list(Small)
    Replacement_names = list(Large)
    #This loop checks to see if the small plan variable name is contained within a large plan variable name. If true, the small plan variable name is replaced with the name of the large plan variable name.
    for r_name in Replacement_names:
        for c_name in Names_to_change:
            if c_name in r_name:
                Small.rename(columns = {c_name:r_name}, inplace = True)
    #This next line renames the variables we renamed earlier, so that they go back to their original name (and aren't duplicated variable names so we can concat!)
    Small = Small.rename(columns = {'x_PLAN_NAME': 'PLAN_NAME', 'x_PLAN_NUM': 'PLAN_NUM', 'x_SPONS_EIN': 'SPONS_EIN', 'x_ADMIN_NAME':'ADMIN_NAME', 'x_ADMIN_PHONE_NUM':'ADMIN_PHONE_NUM', 'x_PREPARER_PHONE_NUM':'PREPARER_PHONE_NUM'})

    #This section puts the large and small plan files for the same year into one dataframe, and then exports it as a csv so we can use it later on.
    BaseForm = pd.concat([Large, Small])
    print(str(year)+"BaseForm opened")

    #Free up memory
    del Small_Chunk, Small, Large_Chunk, Large

    Other_Years = [2009,2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021]
    Other_Years.remove(year)

    for other_year in Other_Years:
          Other_Year = BaseForm[BaseForm['FORM_TAX_PRD'].str.startswith(str(other_year))==True]
          if not Other_Year.empty:
            print('Appending '+str(other_year)+' data found in '+str(year)+' file to '+str(other_year)+' file')
            Chunk = pd.read_csv(path+str(other_year)+"Form_5500.csv", dtype = BaseForm_Dict, chunksize=100000,na_values = '+')
            Current_File = pd.concat(Chunk)
            df = pd.concat([Current_File,Other_Year])
            df.to_csv(path+str(other_year)+'Form_5500.csv', index=False, chunksize=100000)
            print('appended successfully')
            #time.sleep(60)
          del Other_Year

```

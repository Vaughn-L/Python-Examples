```python
#Loop over files in folder
os.chdir(r"C:\Users\Matthew.Vaughn\OneDrive - Berkeley Research Group, LLC\Python Code Examples\PDFs and Images\Test Data\10Ks and 10Qs\PDFs")
output_folder = r"C:\Users\Matthew.Vaughn\OneDrive - Berkeley Research Group, LLC\Python Code Examples\PDFs and Images\Test Data\10Ks and 10Qs\Output"

files = os.listdir()

Phrases = ['average client investment funds', 'period-end client investment funds']

#loop over files
for file in files:
    PDF = open(file,'rb')
    pdfReader = PyPDF2.PdfReader(PDF)
    totalPages = len(pdfReader.pages)

    for page in range(1,totalPages):
        tables = tabula.read_pdf(file, pages=str(page),lattice = True, multiple_tables=True, output_format="json")
        for table in tables:
            new_top = table['top']-30
            df = tabula.read_pdf(file, pages=str(page), stream=True,area=(new_top,table['left'],table['bottom'],table['right']))
            try:
                df = df[0]
                df.replace(' \$','', inplace=True,regex=True)
                df = df.replace({'\$': np.NaN,'%':''}, regex=True)
                df = df.dropna(axis=1,how='all')
                df.rename(columns= {df.columns[0]: "First"}, inplace = True)
                for phrase in Phrases:
                    try:
                        if df['First'].str.contains(phrase).any():
                            name = file.replace('.pdf','')
                            df.to_csv(rf'{output_folder}\{name}_{phrase}_page{str(page)}.csv',index=False)
                    except:
                        pass
            except:
                pass
```

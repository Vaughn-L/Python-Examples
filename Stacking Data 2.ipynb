```python
import pandas as pd
import re
```

```python
all_sheets_dict = pd.read_excel(r'\\na2-fs1\DCDATA6\\Cases\Active\Zydus (049586)\Data\\IQVIA Data (Not for Cleve)\\Topiramate IQVIA Data July 2017-June 2023-c.xlsx', sheet_name=None)
Measures = ['Sales','EUTRx','TRx','Ext']

for sheet_name, sheet_df in all_sheets_dict.items():
    df_output = pd.DataFrame(columns=['Manufacturer','Strength','NDC','Date']) 
    df = sheet_df

    for measure in Measures:
        measure_cols = [col for col in df.columns if col.startswith(measure)]
        measure_melted = pd.melt(df, id_vars = ['Manufacturer','Strength','NDC'], value_vars = measure_cols)
        measure_melted['Date'] = measure_melted['variable'].str.split('\n',expand=True)[1]
        measure_melted[measure] = measure_melted['value']
        measure_melted = measure_melted.drop(['variable','value'], axis=1)
        df_output = pd.merge(df_output, measure_melted, how='outer', on=['Manufacturer','Strength','NDC','Date'])

    df_output.to_excel(r'\\na2-fs1\DCDATA6\\Cases\Active\Zydus (049586)\Analysis\\'+sheet_name+' IQVIA Data July 2017-June 2023_stacked.xlsx', index=False)
```
